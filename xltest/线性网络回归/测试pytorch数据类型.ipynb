```python
import re
```

```python
def read_car_usage_report():
    with open("./car_usage_predict.txt", 'r') as f:
        lines = f.readlines()
    return [re.sub('[^A-Za-z]+', ' ', line).strip().lower() for line in lines]
```

```python
lines = read_car_usage_report()
print(f"# 文本行数： {len(lines)}")
print(lines[0])
print(lines[-2])
```

```python
def tokenize(lines, token='word'):
    if token == 'word':
        return [line.split() for line in lines]
    elif token == 'char':
        return [list(line) for line in lines]
    else:
        print(f'error: undefined token type: {token}')
        
tokens = tokenize(lines)
for i in range(10):
    print(tokens[i])
```

```python
tokens = tokenize(lines, 'char')
for i in range(10):
    print(tokens[i])
```
